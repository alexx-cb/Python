```python
import pandas as pd
import matplotlib.pyplot as plt
# Datos simulados
data = {'Fecha': ['2023-01-01', '2023-01-02', '2023-01-03'],
'Temperatura': [15, 18, 12]}
df = pd.DataFrame(data)
df['Fecha'] = pd.to_datetime(df['Fecha']) # Convertir a fecha

# Gráfico simple
plt.plot(df['Fecha'], df['Temperatura'], marker='o')
plt.title('Temperaturas Diarias')
plt.xlabel('Fecha')
plt.ylabel('Temperatura (°C)')
plt.show()
```
